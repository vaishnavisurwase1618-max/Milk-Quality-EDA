# Problem Statement

The increasing use of Generative AI tools is changing how students learn and complete academic tasks. However, its impact on academic performance, study habits, and learning outcomes is not clearly understood. Therefore, this project aims to analyze student data to understand the effects of AI usage on education.

# Librareis
- NumPy – Used for numerical computations and array operations.
- Pandas – Used for data loading, cleaning, manipulation, and analysis.
- Matplotlib – Used for creating basic charts and visualizations.
- Seaborn – Used for creating attractive statistical visualizations and exploring data patterns.

# Columns and its meaning
- Student_ID: Unique ID given to each student.
- Major_Category: Student's field of study (e.g., Engineering, Business, Science).
- Year_of_Study: Current academic year of the student (1st, 2nd, 3rd, etc.).
- Pre_Semester_GPA: GPA before the semester started.
- Weekly_GenAI_Hours: Number of hours spent using AI tools per week.
- Primary_Use_Case: Main purpose for using AI (e.g., assignments, coding, research, studying).
- Prompt_Engineering_Skill: Student's skill level in writing effective AI prompts.
- Tool_Diversity: Number or variety of AI tools used by the student.
- Paid_Subscription: Whether the student uses a paid AI service (Yes/No).
- Traditional_Study_Hours: Hours spent studying without AI assistance.
- Perceived_AI_Dependency: How much the student feels dependent on AI tools.
- Institutional_Policy: College/university rules regarding AI usage.
- Anxiety_Level_During_Exams: Student's anxiety or stress level during exams.
- Post_Semester_GPA: GPA after the semester ended.
- Skill_Retention_Score: Measure of how well the student remembers and retains learned concepts.
- Burnout_Risk_Level: Risk of mental exhaustion or burnout due to workload and study habits.

# Multi-Variate Analysis

```python
# Select the specific columns for the heatmap
cols_to_plot = ['Tool_Diversity', 'Weekly_GenAI_Hours', 'Post_Semester_GPA']

# Calculate the correlation matrix
correlation_matrix = df[cols_to_plot].corr()

# Create the heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix, 
    annot=True,        # Show the correlation values
    cmap='coolwarm',   # Red-Blue color scheme
    fmt=".2f",         # Format to 2 decimal places
    linewidths=0.5
)
 
plt.title('Correlation Heatmap: AI Usage and GPA')
plt.show()
```

## Observation:
Based on the heatmap:

- **Tool_Diversity and Post_Semester_GPA** have a correlation of 0.02, which is extremely weak positive correlation. This suggests that using a wider variety of AI tools has almost no impact on GPA.
- **Weekly_GenAI_Hours and Post_Semester_GPA** have a correlation of -0.01, which is essentially no correlation. This indicates that the number of hours spent using GenAI tools does not significantly affect students' GPA.
- Overall, all correlation values are very close to 0, suggesting no meaningful linear relationship between AI usage metrics (Tool Diversity and Weekly GenAI Hours) and Post-Semester GPA in this dataset.
